```python
import chromadb
```

```python
# 1. 创建客户端
client = chromadb.Client()
```

```python
# 2. 创建集合
collection = client.create_collection(name="my_collection")
```

```python
# 3. 插入数据
collection.add(
    ids=["1", "2", "3"],
    documents=["Hello world", "Bye world", "Goodbye world"],
)
```

```python
# 4. 查看数据
collection.peek()
```

```python
# 5. 获取数据
collection.get(ids="2")
```

```python
# 6. 相似检索
collection.query(query_texts=["hello world"])
```
